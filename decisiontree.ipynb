```python
x=data.drop("species",axis=1)
y=data["species"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
%matplotlib inline
```

```python
treeclassifier=DecisionTreeClassifier()
```

```python
treeclassifier
```

```python
treeclassifiergini=DecisionTreeClassifier(criterion="gini",max_depth=5,class_weight="balanced",random_state=42)
treeclassifierentropy=DecisionTreeClassifier(criterion="entropy",max_depth=5,class_weight="balanced")
```

```python
treeclassifiergini.fit(
x_train,y_train
)
```

```python
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(treeclassifier,filled=True)
```

```python
treeclassifierentropy.fit(
x_train,y_train)
tree.plot_tree(treeclassifierentropy,filled=True)
```

```python
y_pred=treeclassifiergini.predict(x_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
```

```python
cm=confusion_matrix(y_test,y_pred)
cm
print(classification_report(y_test,y_pred))
```

```python
##gridsearchcv
from sklearn.model_selection import GridSearchCV
parameters={"criterion":["gini","entropy"],"max_depth":[3,5,7],"class_weight":["balanced",None]}
grid=gridsearch=GridSearchCV(estimator=treeclassifierentropy,param_grid=parameters,cv=5)
grid.fit(x_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
from sklearn.metrics import accuracy_score
y_pred=grid.predict(x_test)
accuracy_score(y_test,y_pred)
```
